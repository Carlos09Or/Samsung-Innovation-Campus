# Método append()

# Método extend()

Toma cada uno de los elementos de un iterable (como otra lista, tupla, etc.) y los agrega uno por uno al final de la lista principal (lo que se conoce como aplanar o fusionar).

```python
carros_lujo = ["Ferrari", "Lamborghini"]

# Usando extend() para añadir una nueva lista de carros de lujo
nuevos_carros = ["Mercedes-Benz", "BMW"]
carros_lujo.extend(nuevos_carros)

print(carros_lujo)
```

# Operaciones Básicas de Conjuntos (set)

## Método add()

El método .add() añade un elemento a un conjunto. Si el elemento ya existe, el conjunto no cambia.

```python
carros = {"Toyota", "Honda", "Ford"}
carros.add("Chevrolet")
print(carros)
```

## Método remove()

El método .remove() elimina un elemento; si el elemento no está en el conjunto, se produce un error.

```python
carros = {"Toyota", "Honda", "Ford"}
carros.remove("Toyota")
carros.add("Chevrolet")
print(carros)
```

Los conjuntos son muy útiles para limpiar listas que tienen elementos repetidos.

```python
# Ejemplo de eliminación de duplicados
numeros_duplicados = [1, 2, 2, 3, 4, 4, 5]
conjunto_numeros = set(numeros_duplicados)
print(conjunto_numeros)
```

También son excelentes para operaciones de conjuntos como la unión, la intersección y la diferencia.

## Método Unión

El método .union() (o el operador |) crea un nuevo conjunto que contiene todos los elementos de ambos conjuntos, sin duplicados.

```python
A = {'perro', 'gato', 'loro'}
B = {'gato', 'conejo', 'serpiente'}

union_conjuntos = A.union(B)
print(f"Unión usando el método .union(): {union_conjuntos}")
```

```python
#conjuntos de paises, método unión "|"
paises1 = {"Francia","Argentina","Venezuela","Panamá"}
paises2 = {"India","Finlandia","Iran","Alemania"}
union_paises = paises2 | paises1
print(f"Unión usando el operador |: {union_paises}")
```

## Método Intersección

El método .intersection() (o el operador &) devuelve un nuevo conjunto con los elementos que son comunes a ambos conjuntos.

```python
A = {'perro', 'gato', 'loro'}
B = {'gato', 'conejo', 'serpiente'}

interseccion_conjuntos = A.intersection(B)
print(f"Intersección usando el método .intersection(): {interseccion_conjuntos}")
```

```python
#conjuntos de numeros, método intersección "&"
numeros1 = {1, 2, 6, 4, 5}
numeros2 = {9, 12, 5, 6, 10}
interseccion_numeros = numeros1 & numeros2
print(interseccion_numeros)
```

## Método Diferencia

El método .difference() (o el operador -) devuelve un nuevo conjunto con los elementos que están en el primer conjunto, pero no en el segundo.

```python
A = {'perro', 'gato', 'loro'}
B = {'gato', 'conejo', 'serpiente'}

diferencia_conjuntos = B.difference(A)
print(f"Diferencia A - B (método .difference()): {diferencia_conjuntos}")
```

```python
#conjuntos de carros, método de intersección "-"
carros1 = {"Toyota", "Honda", "Ford"}
carros2 = {"Honda", "Chevrolet", "Ford"}
diferencia_carros = carros1 - carros2
print(diferencia_carros)
```

## ⏱️ Minuto 00 - 15 | Bloque 1: Métodos Avanzados de Diccionarios
Los diccionarios permiten almacenar datos en pares `clave: valor`. Veremos métodos profesionales para gestionar datos sin generar errores:
- `setdefault(clave, valor_por_defecto)`: Inicializa claves automáticamente si no existen.
- `update()`: Modifica múltiples valores o añade nuevos elementos en bloque.
- `pop()` y `popitem()`: Eliminan elementos de forma segura controlando excepciones.

### Celda 1: Método setdefault()

```python
# Celda 1: Uso de setdefault() para inicializar claves automáticamente
inventario = {'manzanas': 10, 'peras': 5}

# 1. Intentamos obtener una clave que ya existe (devuelve su valor actual)
stock_manzanas = inventario.setdefault('manzanas', 0)
print(f"Stock de manzanas: {stock_manzanas}")

# 2. Intentamos obtener una clave que NO existe ('naranjas')
# Como no existe, la añade automáticamente con valor 0 y nos lo devuelve
stock_naranjas = inventario.setdefault('naranjas', 0)
print(f"Stock de naranjas (recién agregada): {stock_naranjas}")
print("Inventario actualizado:", inventario)
```

### Celda 2: Método update()

```python
# Celda 2: Uso de update() para modificar y añadir en bloque
precios_base = {'cafe': 3.50, 'te': 2.00, 'pan': 1.00}
print("Precios originales:", precios_base)

# Actualizamos precios existentes y añadimos un producto nuevo en un solo paso
nuevos_datos = {'cafe': 4.00, 'pan': 1.20, 'jugo': 3.00}
precios_base.update(nuevos_datos)

print("Precios tras el update():", precios_base)
```

### Métodos pop() y popitem()
pop(clave):Elimina una clave específica que tú le indiques y te devuelve su valor asociado.

popitem():Elimina y devuelve el último par (clave-valor) que se haya añadido al diccionario

```python
# Celda 3: Eliminación segura con pop() y popitem()
usuario = {'id': 1024, 'nombre': 'Sofía', 'rol': 'Editor', 'activo': True}
print("Diccionario inicial:", usuario)

# pop(clave) elimina una clave específica y devuelve su valor
rol_eliminado = usuario.pop('rol')
print(f"Clave 'rol' eliminada con valor: {rol_eliminado}")

# popitem() elimina y devuelve el último par (clave, valor) insertado
ultimo_elemento = usuario.popitem()
print(f"Elemento eliminado con popitem(): {ultimo_elemento}")

print("Diccionario final resultante:", usuario)
```

- **`defaultdict`**: Importado de `collections`, previene errores de tipo `KeyError` asignando un tipo de dato por defecto (como `int` o `list`) al invocar claves inexistentes.
- **`datetime` y `math`**: Librerías estándar para manipulación temporal y cálculos matemáticos avanzados.

Diccionario

```python
# Celda 1: Uso de defaultdict para evitar errores de claves inexistentes (KeyError)
from collections import defaultdict

# Creamos un defaultdict que asigna por defecto un entero (0) a cualquier clave nueva
conteo_seguro = defaultdict(int)

secuencia = "python"
for letra in secuencia:
    conteo_seguro[letra] += 1

print("Conteo seguro de letras:", dict(conteo_seguro))
```

```python
from collections import defaultdict

texto = "Banana"
# Le pasamos int como fábrica; el valor por defecto de un int es 0
conteo = defaultdict(int)

for letra in texto:
    conteo[letra] += 1  # Si la letra no existe, crea la clave con valor 0 y luego suma 1

print(conteo)  # defaultdict(<class 'int'>, {'b': 1, 'a': 3, 'n': 2})
```

listas


```python
from collections import defaultdict

pares = [("rojo", "manzana"), ("amarillo", "plátano"), ("rojo", "fresa"), ("amarillo", "limón")]

# Usamos list como fábrica
agrupados = defaultdict(list)

for color, fruta in pares:
    agrupados[color].append(fruta)  # Si el color no existe, crea una lista vacía y añade la fruta

print(dict(agrupados))
# Resultado: {'rojo': ['manzana', 'fresa'], 'amarillo': ['plátano', 'limón']}
```

## ⏱️ Minuto 25 - 40 | Bloque 3: Tipos de Datos Conjunto (`set`)
Los conjuntos (`set`) simulan las estructuras matemáticas:
- Almacenan elementos **únicos** (eliminan duplicados automáticamente).
- No tienen un orden fijo y **no admiten indexación ni slicing**.
- Permiten operaciones potentes: Unión (`|`), Intersección (`&`), Diferencia (`-`) y Diferencia Simétrica (`^`).

```python
# Celda 1: Los conjuntos eliminan duplicados solos
# Supongamos una lista de correos o IDs repetidos
asistentes_repetidos = ['Ana', 'Carlos', 'Ana', 'Sofia', 'Carlos', 'David']

# Convertimos la lista en un conjunto (set) usando llaves {} o la función set()
asistentes_unicos = set(asistentes_repetidos)

print("Lista original con repetidos:", asistentes_repetidos)
print("Conjunto limpio (sin duplicados):", asistentes_unicos)
```

### Celda 2: Operación de Unión (|)

```python
# Celda 2: Unión de conjuntos (|)
# Une los elementos de ambos grupos, eliminando los repetidos automáticamente.
grupo_manana = {'Ana', 'Carlos', 'Sofia'}
grupo_tarde = {'Sofia', 'David', 'Elena'}

# Unimos los dos grupos con el símbolo barra vertical (|)
todos_los_estudiantes = grupo_manana | grupo_tarde

print("Estudiantes de la mañana:", grupo_manana)
print("Estudiantes de la tarde:", grupo_tarde)
print("Unión (Todos los estudiantes sin repetir):", todos_los_estudiantes)
```

### Celda 3: Operación de Intersección (&)

```python
# Celda 3: Intersección de conjuntos (&)
# Muestra únicamente los elementos que se repiten en ambos grupos.
curso_python = {'Ana', 'Erick', 'David', 'Elena'}
curso_ia = {'David', 'Elena', 'Jhoandry', 'Gabriela'}

# Buscamos quiénes están en los dos cursos a la vez con el símbolo ampersand (&)
ambos_cursos = curso_python & curso_ia

print("Estudiantes que están en ambos cursos:", ambos_cursos)
```

### Celda 4: Operación de Diferencia (-) y Diferencia Simétrica (^)

```python
# Celda 4: Diferencia (-) y Diferencia Simétrica (^)
# - Diferencia (-): Muestra los elementos que están en el primer conjunto pero NO en el segundo.
# - Diferencia simétrica (^): Muestra los elementos que están en uno o en otro, pero NO en ambos.

equipo_rojo = {'Ana', 'Carlos', 'Sofia'}
equipo_azul = {'Sofia', 'David', 'Elena'}

# Quiénes están solo en el equipo rojo (restándole los del azul)
solo_rojos = equipo_rojo - equipo_azul
print("Estudiantes que están exclusivamente en el equipo rojo:", solo_rojos)

# Quiénes están en un equipo o en el otro, pero no compartidos
exclusivos = equipo_rojo ^ equipo_azul
print("Estudiantes que no se repiten entre ambos equipos:", exclusivos)
```

## ⏱️ Minuto 40 - 55 | Bloque 4: Listas Bidimensionales (Matrices)
Una lista bidimensional es una lista que contiene otras listas en su interior, modelando una cuadrícula o tabla estructurada en **filas y columnas**.
- **Acceso:** `matriz[fila][columna]`
- **Recorrido:** Se emplean bucles `for` anidados.

### **LISTAS BIDIMENSIONALES**

### Las listas bidimensionales son listas anidadas dentro de otra lista, creando una estructura de datos que se asemeja a una tabla o una matriz Cada lista interna (o sub-lista) representa una fila, y los elementos dentro de cada sub-lista representan las columnas.

```python
# Lista unidimensional (una fila)
lista_simple = [1, 2, 3, 4]

# Lista bidimensional (una matriz 3x3)
matriz = [
    [1, 2, 3],  # Fila 0
    [4, 5, 6],  # Fila 1
    [7, 8, 9]   # Fila 2
]
print(lista_simple)
print(matriz)
```

```python
lista_frutas = [
    ["manzana", "banana", "cereza"],
    ["mango", "kiwi", "naranja"],
    ["pera", "uva", "sandia"]
    ]

print(lista_frutas[1][0])
```

```python
# Creación de una matriz de 3x3
matriz = [[10, 20, 30], [40, 50, 60], [70, 80, 90]]

print(
    "Elemento específico en Fila 1, Columna 2:", matriz[1][2]
)  # Imprime 60

# Recorrido completo formateado como tabla
print(
    "--- Visualización tabular de la matriz ---"
)
for fila in matriz:
    for elemento in fila:
        print(elemento, end="\t")
    print()  # Salto de línea por cada fila

# Modificación dinámica de un valor
matriz[0][0] = 999
print("Matriz tras modificar la posición [0][0]:", matriz[0])
```

### **tic tac toe**

```python
# El jugador 'X' hace un movimiento
# Posición:(índices 1, 0)
tablero = [
    [' ', ' ', ' '],
    [' ', ' ', ' '],
    [' ', ' ', ' ']
]
tablero[1][1] = 'X'

# Imprimimos el tablero actualizado para ver el cambio
for fila in tablero:
    print(fila)
```

```python
#Jueguito
# Tablero inicial vacío
tablero = [[' ', ' ', ' '], [' ', ' ', ' '], [' ', ' ', ' ']]

# Jugadas alternadas (X y O)
tablero[0][2] = 'X'
tablero[1][2] = 'O'
tablero[2][2] = 'X'
tablero[0][2] = 'O'

# Mostrar tablero final
for fila in tablero:
    print(fila)
```

⏱️ Minuto 55 - 60 | Reto Práctico: El Buscador de Tesoros Únicos

Reto de Programación en Parejas: Tienes un tablero (matriz) donde hay diferentes premios escondidos (letras o palabras). Tu misión es escribir un script que recorra la matriz, meta todos los premios en un conjunto (set) para eliminar los repetidos, y te muestre cuáles fueron los premios únicos que se encontraron en el mapa.

image.png

### Respuesta

```python
# Matriz que representa el tablero del juego con premios repetidos
tablero_premios = [
    ['oro', 'plata', 'oro'],
    ['bronce', 'diamante', 'plata'],
    ['oro', 'bronce', 'diamante']
]

# Creamos un conjunto (set) vacío para guardar los premios únicos
premios_unicos = set()

# Recorremos la matriz fila por fila, elemento por elemento
for fila in tablero_premios:
    for premio in fila:
        # Añadimos cada premio al conjunto (los duplicados se eliminan solos)
        premios_unicos.add(premio)

# Mostramos el resultado final
print("💎 ¡Premios únicos encontrados en todo el tablero:")
print(premios_unicos)
```
